### ✏️ Atividade Prática 2-A — Sua vez de programar

Dado um `preco` de sua escolha, calcule o preço com **10% de desconto**.

```python
# Preencha com um preço de sua escolha
preco_original = 39.78

desconto = preco_original * 0.1
preco_final = preco_original - desconto

print(f"Preço original: R$ {preco_original:.2f}")
print(f"Desconto (10%): R$ {desconto:.2f}")
print(f"Preço com desconto: R$ {preco_final:.2f}")
```

### ✏️ Atividade Prática 2-D — Sua vez de programar

Uma loja vendeu `quantidade` unidades a `preco_unitario` cada. Calcule: total da venda, **comissão do vendedor (5%)** e **lucro líquido**.

```python
# Preencha com quantidade e preço unitário
quantidade = 5
preco_unitario = 119.99

total_venda = quantidade * preco_unitario
comissao = total_venda * 0.05
lucro_liquido = total_venda - comissao

print(f"Total da venda:  R$ {total_venda:.2f}")
print(f"Comissão (5%):   R$ {comissao:.2f}")
print(f"Lucro líquido:   R$ {lucro_liquido:.2f}")
```

✏️ Atividade Prática 5-D — Sua vez de programar

Use while para fazer uma contagem regressiva de 10 até 1. Ao final, imprima 'Lançar!'.

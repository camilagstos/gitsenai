✏️ Atividade 1 — Contar por setor (CSV): usando cadastro_funcionario.csv, filtre e conte quantos funcionários existem no setor "Estoque". 

```python
import csv

with open("cadastro_funcionario.csv", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    funcionarios = list(leitor)
    
funcionarios_estoque = [f for f in funcionarios if f["setor"] == "Estoque"]

print(f" Funcionários no departamento de estoque: {len(funcionarios_estoque)}")
```

✏️ Atividade 2 — Escrever um resumo (JSON): a partir do CSV, monte um dicionário contando quantos funcionários existem em cada setor e salve esse resumo em um novo arquivo relatorio_setores.json.

```python
import csv
import json

contagem_setores = {}

with open("cadastro_funcionario.csv", encoding="utf-8") as arquivo:
    funcionarios = list(csv.DictReader(arquivo))

for funcionario in funcionarios:
    setor = funcionario["setor"]
    contagem_setores[setor] = contagem_setores.get(setor, 0) + 1

with open("relatorio_setores.json", "w", encoding="utf-8") as arquivo_json:
    json.dump(contagem_setores, arquivo_json, ensure_ascii=False, indent=4)

print(contagem_setores)
```

✏️ Atividade 6 — Atraso de outro funcionário: calcule o atraso (em minutos) do mesmo registro da Atividade 5, comparando com a entrada prevista das 08:00. Se o resultado for negativo, o que isso significa?

```python
from openpyxl import load_workbook
from datetime import datetime

arquivo = load_workbook("recursos_humanos.xlsx")
aba = arquivo["banco_horas"]

id_funcionario = 7
data_procurada = datetime(2023, 1, 2).date()

for linha in aba.iter_rows(min_row=2, values_only=True):

    if linha[0] == id_funcionario and linha[3].date() == data_procurada:

        entrada_prevista_min = 8 * 60
        entrada_real = linha[4]

        entrada_real_min = entrada_real.hour * 60 + entrada_real.minute

        atraso_minutos = entrada_real_min - entrada_prevista_min

        print(f"Atraso: {atraso_minutos} minutos.")
        
# Caso o resultado fosse negativo, significa que o funcionário chegou antes do horário previsto.
```

✏️ Atividade 10 — calcular_atraso com parâmetro padrão: crie a função (como no exemplo) e teste uma vez com a entrada prevista padrão (08:00) e outra vez com um horário customizado (ex.: turno que começa às 07:00).

```python
def calcular_atraso(hora_entrada, minuto_entrada, hora_prevista=8, minuto_previsto=0):
    minutos_reais = hora_entrada * 60 + minuto_entrada
    minutos_previstos = hora_prevista * 60 + minuto_previsto
    return minutos_reais - minutos_previstos

print(calcular_atraso(8, 30))        
print(calcular_atraso(7, 45, hora_prevista=7))
```

✏️ Atividade 11 — validar_email: transforme a checagem de regex da Seção 2 em uma função validar_email(email) que retorna True ou False.

```python
import re

def validar_email(email):
    padrao = r"^[^\s@]+@[^\s@]+\.[^\s@]+$"

    if re.match(padrao, email):
        return True
    else:
        return False

print(validar_email("camila.gomes@grupoalfa.com.br"))
print(validar_email("camila.gomes@@grupoalfa.com.br"))
```

✏️ Atividade 12 — classificar_pontualidade: crie a função (como no Exemplo 2) e teste com 3 valores diferentes de atraso.

```python
def classificar_pontualidade(atraso_minutos, tolerancia=5):
    if atraso_minutos <= tolerancia:
        return "Pontual"
    return "Atrasado"

print(classificar_pontualidade(2))          
print(classificar_pontualidade(25))         
print(classificar_pontualidade(14, tolerancia=15)) 
```

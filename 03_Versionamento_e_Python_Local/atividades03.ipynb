✏️ Atividade 3-E (no notebook): a fila de atendimento abaixo precisa chamar a primeira pessoa (ela é a próxima a ser atendida). Use .pop(0) para retirar essa pessoa da fila, guarde o nome numa variável (você vai usar esse valor na mensagem) e exiba quem foi chamado e como ficou a fila.

```python
fila = ["Marcos", "Beatriz", "Rafael"]

# Retire a PRIMEIRA pessoa da fila usando .pop(0) e guarde numa variável
proximo = fila.pop(0)

print(f"Chamando: {proximo}")
print(fila)
```

✏️ Atividade Prática 3-D (no notebook): crie um dicionário representando um animal de estimação (nome, espécie, idade) e exiba uma frase completa.

```python
animal = {
    "nome":    "Thor",
    "especie": "Cachorro",
    "idade": 6}

print(f"Meu animal se chama {animal['nome']} e tem {animal['idade']} anos.")

# Bônus: atualize a idade para um ano a mais
animal["idade"] = animal["idade"] + 1
print(f"Ano que vem terá {animal['idade']} anos.")
```
